```python
# Importar a Biblioteca
import pandas as pd
```

#### **IMPORTANDO A PLANILHA E OBSERVANDO POSSÍVEIS DADOS QUE SERÃO TRATADOS**

```python
# Abrir a planilha pelo pandas e criar um DataFrame
df = pd.read_csv(
    "vendas_projeto_auditor.csv",
    sep=",",
    encoding="utf-8-sig"
)

print("Planilha csv importada com sucesso!")

print("\nExibir 5 primeiras linhas")
display(df.head())


print(f"\nA quantidade de linhas e colunas respectivamente é: {df.shape}")
```

#### **VERIFICAR A QUANTIDADE DE LINHAS VAZIAS POR COLUNAS**

```python
print("\nTotal de células vazias por colunas: ")
display(df.isna().sum().sort_values(ascending=False))
```

#### **VERIFICAR A QUANTIDADE DE CÉLULAS DUPLICADAS E QUAIS SÃO**

```python
duplicadas_exatas = df[df.duplicated()]

print("\nTodos os registros duplicados encontrados: ")
display(duplicadas_exatas)

total_duplicadas = df.duplicated().sum()

print(f"A planilha possui {total_duplicadas} linhas duplicadas.")
print(40 * "_")

ids_repetidos = (

    df[
        df.duplicated(
            subset="id_venda",
            keep=False
        )
    ]
    .sort_values(by="id_venda")
)
print("\nVendas apenas com IDs repetidos: ")
display(ids_repetidos)

qtd_precos_dif_mesmo_id = df.groupby("id_venda")[["quantidade", "preco_unitario"]].nunique()
print("\nVerificar quantidades diferentes e preços diferentes em relaçao aos mesmos IDs:")
display(qtd_precos_dif_mesmo_id )
print("\nNão existem quantidades diferentes e preços diferentes relacionados aos mesmos IDs.")
print(80 * "_")
```

#### **REMOÇÃO DAS CÉLULAS DUPLICADAS**

```python
# Copia do DataFrame
df_limpo = df.copy()

# Remover Duplicadas
df_limpo = df_limpo.drop_duplicates()
print("\nLinhas originais:", len(df))
print("\nLinhas após limpeza:", len(df_limpo))

total_duplicadas_att = df_limpo.duplicated().sum()
print(f"\nAgora o número de células duplicadas no DataFrame atualizado é de: {total_duplicadas_att}.")

# Tratamento Células Vaziar
df.groupby("produto")["preco_unitario"].nunique()
```

#### **PADRONIZAR A FORMATAÇÃO DE TEXTO DA COLUNA DE PRODUTOS**

```python
df_limpo["produto"] = (
    
    df_limpo["produto"]
    .str.strip()
    .str.title()
)
print("\nOs produtos foram padronizado com primeira letra maiúscula e foram removidos os espaços extras.")
display(df_limpo["produto"])

print("\nQuantidade de vezes que um produto aparece na planilha:")
display(df_limpo["produto"].value_counts(dropna=False))
```

#### **TRATAMENTO DAS CÉLULAS VAZIAS DA COLUNA DE PREÇO**

```python
# Verificar se existe preços diferentes de um mesmo produto
preco_dif_produto = df_limpo.groupby("produto")["preco_unitario"].nunique()
display(preco_dif_produto)
print("\nA tabela atualizada não possui nenhum valor diferente de um mesmo produto.")
print(74 * "_")

# Verificar preço de cada produto
precos_produtos = df_limpo.groupby("produto")["preco_unitario"].first()
print("\nTABELA DE PREÇOS POR PRODUTO")
display(precos_produtos)

# Criar tabela de referência para preencher os preços vazios
tabela_referencia = (
    df_limpo.groupby
    ("produto")
    ["preco_unitario"]
    .transform("first")
)

# Preencher preços vazios através da tabela de referência
verificar_precos_vazios_antes = (
    df_limpo["preco_unitario"]
    .isna()
    .sum()
)

# Preencher preços vazios com a tabela de referência
df_limpo["preco_unitario"] = df_limpo["preco_unitario"].fillna(tabela_referencia)

# Verificar quantos preços continuam vazios depois da limpeza
verificar_precos_vazios_depois = (
    df_limpo["preco_unitario"]
    .isna()
    .sum()
)

# Calcular quantos preços foram preenchidos
precos_preenchidos = verificar_precos_vazios_antes - verificar_precos_vazios_depois

if verificar_precos_vazios_depois == 0:
    print("\nTodos os preços foram preenchidos com sucesso!")
    print(f"Preços preenchidos nesta execução: {precos_preenchidos}")
    print(55 * "_")
    
    print("\nTabela atualizada - ID, Produto e Preço:")
    display(df_limpo[["id_venda","produto","preco_unitario"]])

else:
    print("\nAtenção: Ainda há células vazias presentes nessa coluna")
    print(f"Preços preenchidos: {precos_preenchidos}")
    print(f"Preços ainda vazios: {verificar_precos_vazios_depois}")
```

#### **TRATAMENTO DAS CÉLULAS VAZIAS e QUANTIDADES NEGATIVAS DA COLUNA DE QUANTIDADE**

```python
# Verificar a quantidade de células vazias
total_vendas_sem_qtd = df_limpo["quantidade"].isna().sum()
print(f"A quantidade de vendas que não possuem quantidade são: {total_vendas_sem_qtd}")
print(57 * "_")

# Identificar vendas com quantidades negativas
vendas_sem_qtd = df_limpo[
    df_limpo["quantidade"].isna()
]
print("\nCélulas que estão sem a quantidade:")
display(vendas_sem_qtd[["id_venda", "produto", "quantidade"]])

# Verificar o total de vezes que a quantidade aparece de forma negativa
total_vendas_negativas = (df_limpo["quantidade"] < 0).sum()
print(f"\nVezes que aparecem quantidades negativas na planilha: {total_vendas_negativas}")
print(55 * "_")

# Verificar quais células estão com a quantidade vazia
vendas_negativas = df_limpo[
    df_limpo["quantidade"] 
    < 0
]
print("\nCélulas que apresentam quantidade negativa:")
display(vendas_negativas)

def classificar_quantidade(quantidade):

    if pd.isna(quantidade):
        return "Ausente - REVISAR"

    elif quantidade < 0:
        return "Troca/Devolução"

    elif quantidade == 0:
        return "Zero - REVISAR"

    else:
        return "Venda Válida"

# Aplicar a função e criar coluna status para as vendas
df_limpo["status"] = df_limpo["quantidade"].apply(classificar_quantidade)

status_vazios = df_limpo["status"].isna().sum()

if status_vazios == 0:
    print("\nSUCESSO: Todos os status foram preenchidos!")
    print("Tabela atualizada - ID, Quantidade e Status")
    display(df_limpo[["id_venda", "quantidade", "status"]])

else:
    print("\nATENÇÃO: Existem status não preenchidos!")
    print(f"Quantidade de células vazias: {status_vazios}")

resumo_movimentacoes = df_limpo["status"].value_counts()
print("\n RESUMO DAS MOVIMENTAÇÕES:")
display(resumo_movimentacoes)
```

#### **CALCULO DO FATURAMENTO BRUTO E ANÁLISE FINANCEIRA DAS TROCAS/DEVOLUÇÕES**

```python
# Criação da coluna valor total calcuando o valor total das vendas
df_limpo["valor_total"] = df_limpo["quantidade"] * df_limpo["preco_unitario"]
print("\nVALOR TOTAL DAS MOVIMENTAÇÕES")
display(df_limpo[
    ["produto", "quantidade",
     "preco_unitario", "status", "valor_total"]
])

# Gerar Faturamento Bruto
vendas_validas = df_limpo[
    df_limpo["status"] == "Venda Válida"
]

faturamento_bruto = vendas_validas["valor_total"].sum()

# Função para formatar o valor em R$
def formatar_real(valor):
    sinal = "-" if valor < 0 else ""

    numero = (
    f"{abs(valor):,.2f}"
    .replace(",", "X")
    .replace(".", ",")
    .replace("X", ".")
)
    return f"{sinal}R$ {numero}"

print(f"\nFaturamento bruto: {formatar_real(faturamento_bruto)}")
print(32 * "_")

# Gerar valor das trocas/devoluções
troca_devolucao = df_limpo[
    df_limpo["status"] == "Troca/Devolução"
]

valor_devolucoes = troca_devolucao["valor_total"].sum()
print(f"\nTrocas/Devoluções: {formatar_real(valor_devolucoes)}")
```

#### **REGISTROS QUE PRECISAM DE REVISÃO**

```python
# Criar condição para células que preicsam de revisão
condicao = (
    (df_limpo["status"] == "Ausente - REVISAR") 
    | 
    (df_limpo["status"] == "Zero - REVISAR")
)
display(df_limpo[condicao])
print(f"\nQuantidade de células que precisam de revisão: {condicao.sum()}")
```

```python
# Verificar IDs duplicados após o tratamento
verf_id_tratamento = (
    df_limpo["id_venda"]
    .duplicated()
    .sum()
)

if verf_id_tratamento == 0:
    print("\nSUCESSO: Nenhum ID duplicado encontrado na base tratada.")

else:
    print(f"\nATENÇÃO: {verf_id_tratamento} ocorrências de IDs duplicados na base tratada.")

    # Identificar todos os registros com IDs repetidos
    ids_duplicados = df_limpo[
        df_limpo["id_venda"].duplicated(keep=False)
    ]

    print("\nREGISTROS COM IDs DUPLICADOS:")
    display(ids_duplicados.sort_values(by="id_venda"))
```

#### **MÉTRICAS DE VENDAS: QUANTIDADES, PRODUTOS, VALORES, FORMAS DE PAGAMENTO**

```python
# Contar quantos registros de vendas válidas
qtd_vendas_validas = len(vendas_validas)
print(f"\nA quantidade de registros de vendas válidas: {qtd_vendas_validas}")

# Contar quantos registros de vendas ausentes
qtd_troca_dev = len(troca_devolucao)
print(f"\nA quantidade de registros de vendas de troca/devolução: {qtd_troca_dev}")

# Contar quantos registros de vendas que precisam de revisão
qtd_pendentes = condicao.sum()
print(f"\nA quantidade de registros que precisam ser revisados: {qtd_pendentes}")

# Total de registros da planilha tratada
total_registros = len(df_limpo)
print(f"\nTotal de registros da planilha: {total_registros}")

total_classificados = (
    qtd_vendas_validas
    + qtd_troca_dev
    + qtd_pendentes
)

if total_classificados == total_registros:
    print("SUCESSO: Todos os registros foram contabilizados!")

else:
    print("ATENÇÃO: Existe uma divergência na contagem dos registros.")

# Caclular o ticket médio
ticket_médio = vendas_validas["valor_total"].mean()
print(f"\nValor do ticket médio: R$ {formatar_real(ticket_médio)}")

# Quantidade total de unidades vendidas
total_unidades = vendas_validas["quantidade"].sum()
print(f"\nTotal de unidades vendidas: {total_unidades:.0f}")

# Ranking dos produtos mais vendidos
produtos_mais_vendidos = (
    vendas_validas
    .groupby("produto")["quantidade"]
    .sum()
    .sort_values(ascending=False)
) 
print("\nQuantidade dos produtos mais vendidos:")
display(produtos_mais_vendidos)

# O produto mais vendido com a quantidade
produto_campeao = produtos_mais_vendidos.idxmax()
quantidade_campeao = produtos_mais_vendidos.max()
print(f"\nPRODUTO MAIS VENDIDO: {produto_campeao}")
print(f"QUANTIDADE VENDIDA: {quantidade_campeao:.0f} unidades")

# Produtos com maior faturamento
faturamento_produtos = (
    vendas_validas
    .groupby("produto")["valor_total"]
    .sum()
    .sort_values(ascending=False)
)
print("\nFATURAMENTO POR PRODUTO:")
display(
    faturamento_produtos
    .reset_index(name="Faturamento")
    .style.format(
        {"Faturamento": "R$ {:,.2f}"},
        decimal=",",
        thousands="."
    )
)

# O produto com maior faturamento e valor
produto_faturamento_campeao = faturamento_produtos.idxmax()
valor_faturamento_campeao = faturamento_produtos.max()
print(f"\nPRODUTO COM MAIOR FATURAMENTO: {produto_faturamento_campeao}")
print(f"VALOR DO PRODUTO COM MAIOR FATURAMENTO: {formatar_real(valor_faturamento_campeao)}")

# Quantidade e valores de cada forma de pagamento
print("\nFATURAMENTO E QUANTIDADE POR FORMAS DE PAGAMENTO:")
resumo_pagamentos = (
    vendas_validas
    .groupby("forma_pagamento", dropna=False)
    .agg(
        qtd_registros=("id_venda", "size"),
        faturamento=("valor_total", "sum")
    )
    .sort_values("faturamento", ascending=False)
)
display(
    resumo_pagamentos
    .style.format(
        {"faturamento": "R$ {:,.2f}"},
        decimal=",",
        thousands="."
    )
)

# Forma de pagamento com maior faturamento e valor
forma_pag_campeao = resumo_pagamentos["faturamento"].idxmax()
valor_forma_pag_camepao = resumo_pagamentos["faturamento"].max()
print(f"\nA FORMA DE PAGAMENTO COM MAIOR FATURAMENTO: {forma_pag_campeao}")
print(f"VALOR DA FORMA DE PAGAMENTO COM MAIOR FATURAMENTO: {formatar_real(valor_forma_pag_camepao)}")

# Faturamento por dia
datas_convertidas = pd.to_datetime(
    df_limpo["data"],
    format="%Y-%m-%d",
    errors="coerce"
)
df_limpo["data"] = datas_convertidas
vendas_validas = df_limpo[
    df_limpo["status"] == "Venda Válida"
]

faturamento_diario = (
    vendas_validas.groupby("data")["valor_total"]
    .sum()
    .sort_index()
)
print("\nFATURAMENTO DIÁRIO")
display(faturamento_diario)
```

#### **MÉTRICAS DIA/MÊS**

```python
# Faturamento por dia
datas_convertidas = pd.to_datetime(
    df_limpo["data"],
    format="%Y-%m-%d",
    errors="coerce"
)
df_limpo["data"] = datas_convertidas
vendas_validas = df_limpo[
    df_limpo["status"] == "Venda Válida"
]

faturamento_diario = (
    vendas_validas.groupby("data")["valor_total"]
    .sum()
    .sort_index()
)
print("\nFATURAMENTO DIÁRIO")
display(
    faturamento_diario
    .to_frame(name="Faturamento")
    .head(10)
    .style.format(
        {"Faturamento": "R$ {:,.2f}"},
        decimal=",",
        thousands=".",
        na_rep="—"
    )
)

# Faturamento por mês
faturamento_mensal = (
    vendas_validas
    .groupby(vendas_validas["data"].dt.to_period("M"))["valor_total"]
    .sum()
    .sort_index()
)
print("\nFATURAMENTO MENSAL")
display(
    faturamento_mensal
    .to_frame(name="Faturamento")
    .style.format(
        {"Faturamento": "R$ {:,.2f}"},
        decimal=",",
        thousands=".",
        na_rep="—"
    )
)

# Dia com maior faturamento
dia_campeao = faturamento_diario.idxmax()
valor_dia_campeao = faturamento_diario.max()
print(f"\nDIA DE MAIOR FATURAMENTO: {dia_campeao.strftime('%d/%m/%Y')}")
print(f"FATURAMENTO DO DIA: {formatar_real(valor_dia_campeao)}")

# Mês com maior faturamento
mes_campeao = faturamento_mensal.idxmax()
valor_mes_campeao = faturamento_mensal.max()
print(f"\nMÊS DE MAIOR FATURAMENTO: {mes_campeao}")
print(f"FATURAMENTO DO MÊS: {formatar_real(valor_mes_campeao)}")
```

#### **VALIDAÇÃO DOS INDICADORES**

```python
total_diario = faturamento_diario.sum()
total_mensal = faturamento_mensal.sum()

# Exibir os resultados em reais
print("\nVALIDAÇÃO DOS FATURAMENTOS")

print(f"Faturamento bruto: {formatar_real(faturamento_bruto)}")
print(f"Total diário: {formatar_real(total_diario)}")
print(f"Total mensal: {formatar_real(total_mensal)}")

# Validação dos faturamentos
bruto = round(faturamento_bruto, 2)
diario = round(total_diario, 2)
mensal = round(total_mensal, 2)

if bruto == diario == mensal:
    print("\nSUCESSO: Todos os faturamentos estão consistentes!")

else:
    print("\nATENÇÃO: Divergência encontrada nos faturamentos!")
    print("Revise os seguintes resultados:")

    if bruto != diario:
        diferenca_diaria = bruto - diario

        print("\nDIVERGÊNCIA NO FATURAMENTO DIÁRIO")
        print(f"Faturamento bruto: {formatar_real(bruto)}")
        print(f"Faturamento diário: {formatar_real(diario)}")
        print(f"Diferença: {formatar_real(diferenca_diaria)}")

    if bruto != mensal:
        diferenca_mensal = bruto - mensal

        print("\nDIVERGÊNCIA NO FATURAMENTO MENSAL")
        print(f"Faturamento bruto: {formatar_real(bruto)}")
        print(f"Faturamento mensal: {formatar_real(mensal)}")
        print(f"Diferença: {formatar_real(diferenca_mensal)}")

    if diario != mensal:
        print("\nATENÇÃO: Os totais diário e mensal também são diferentes.")

    print("\nRECOMENDAÇÃO: Verifique as datas, os valores ausentes e os agrupamentos.")
```

#### **RESUMO EXECUTIVO**

```python
resumo_executivo = pd.DataFrame([{
    "total_registros": len(df_limpo),
    "vendas_validas": len(vendas_validas),
    "trocas_devolucoes": len(troca_devolucao),
    "pendentes_revisao": int(qtd_pendentes),

    "faturamento_bruto": faturamento_bruto,
    "valor_devolucoes": valor_devolucoes,
    "ticket_medio": vendas_validas["valor_total"].mean(),
    "unidades_vendidas": int(total_unidades),

    "produto_mais_vendido": produtos_mais_vendidos.idxmax(),
    "produto_maior_faturamento": faturamento_produtos.idxmax(),

    "pagamento_mais_utilizado": resumo_pagamentos["qtd_registros"].idxmax(),
    "pagamento_maior_faturamento": resumo_pagamentos["faturamento"].idxmax(),

    "dia_maior_faturamento": faturamento_diario.idxmax(),
    "mes_maior_faturamento": str(faturamento_mensal.idxmax())
}])


# Organizar os indicadores verticalmente
resumo_visual = resumo_executivo.T.rename(
    columns={0: "Resultado"}
)

# Identificar os indicadores financeiros
indicadores_monetarios = [
    "faturamento_bruto",
    "valor_devolucoes",
    "ticket_medio"
]

# Exibir o resumo com formatação em reais
print("\nRESUMO EXECUTIVO DO AUDITOR INTELIGENTE")

display(
    resumo_visual.style.format(
        formatar_real,
        subset=pd.IndexSlice[
            indicadores_monetarios,
            ["Resultado"]
        ]
    )
)
```
